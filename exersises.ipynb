**Valid Sudoku**

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
import numpy as np
```

```python
A = np.sum(s, axis=1)
for i in A:
  if i != 45:
    print("Not a valid sudoku")
    break
else:
  print("Valid sudoku for rows")
```

**General Questions**

```python
B = np.sum(s, axis=0)
for i in B:
  if i == 45:
    print("Valid sudoku for columns")
    break
else:
  print("Not a valid sudoku")
```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 75, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
ave_age = np.mean(data[0:6,0])
print(f"Average age of students : {ave_age}")
```

```python
#Extract Math marks of all students.
Marks = data[0:6,1]
print(f"Maths marks of all students: {Marks}")
```

```python
#Find the highest Science mark.
highestScienceMark = np.max(data[0:6,2])
print(highestScienceMark)
```

```python
#Get details of the student who scored more than 90 in Math.
N = data[0:6,1] > 90
print(data[N])
```

```python
#Increase Math marks of all students by 5.
print(f"Before : {data[0:6,1]}")
increased = data[0:6,1]+5
print(f"After: {increased}")
```

```python
#Find how many students are younger than 19.
age = data[0:6,0]
#
count = 0
arr = []
for i in age:
  if i < 19:
    count += 1
    arr.append(i)
print(f"There are {count} students younger than 19 : {np.array(arr)}")
```

```python
#Calculate the average marks in each subject (column-wise mean).
averageInMaths = np.mean(data[0:6,1])
averageInScience = np.mean(data[0:6,2])
print(f"Average marks in Maths : {averageInMaths}\nAverage marks in Science : {averageInScience}")
```

```python
#Get data of students who scored at least 80 in both subjects.
data[(data[:,1] >= 80) & (data[:,2] >= 80)]
```
