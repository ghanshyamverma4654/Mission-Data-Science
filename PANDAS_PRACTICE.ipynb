```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('Countries.csv')
```

```python
#Which country has the highest population
data[data['population'] == data['population'].max()][['country','population']]
```

```python
#What is the capital of country with highest population
data[data['population'] == data['population'].max()][['country','population','capital_city']]
```

```python
#Which country has least population
data[data['population'] == data['population'].min()][['country','population','capital_city']]
```

```python
data.head()
```

```python
#Top 5 countries with highest democratic score
data.sort_values(by='democracy_score',ascending = False).head()[['country','capital_city','democracy_score']]
```

```python
#How many regions are there
data['region'].nunique()
```

```python
#How many countries lie in Eastern Europe region
data[data['region'] == 'Eastern Europe'][['country','capital_city']]
```

```python
#Who is the political leader of 2nd highest populated country
data[data['population'] == data['population'].nlargest(2).iloc[1]][['country','capital_city','political_leader']]
```

```python
#How many countries are there whose political leader are unknown
data[data['political_leader'].isna()]['country'].count()
```

```python
#How many countries have Republic in their full name
count = 0
def counting(txt):
  global count
  if 'republic' in txt.lower():
    count += 1
  return txt

data['country_long'] = data['country_long'].apply(counting)
print(f"There are {count} countries which have Republic in their names.")
```

```python
#Which country in African region has highest population
africa_data = data[data['continent'] == 'Africa']
```

```python
africa_data[africa_data['population'] == africa_data['population'].max()][['country','population']]
```
