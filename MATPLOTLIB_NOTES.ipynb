**SEABORN**

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Loading dataset from seaborn's dataset repo
df = sns.load_dataset('tips')
```

```python
df
```

```python
sns.histplot(df['total_bill'], kde = True)
```

```python
sns.histplot(df['total_bill'],bins = 50,color="orange",kde=True)
```

```python
#dist/histplot
plt.subplot(1,2,1)
sns.histplot(df["total_bill" ], bins = 50, kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'], bins= 50, kde = True)
```

```python
#Jointplot
sns.jointplot(x='total_bill', y='tip', data = df, kind = 'resid')
#Kind = scatter/kde/hist/hex/reg/resid
```

```python
#Pairplot
sns.pairplot(df,hue='size',palette = 'rainbow')
```

```python
#Rugplot
sns.scatterplot(x='total_bill',y='tip',data=df)
sns.rugplot(x='total_bill',y='tip',data=df, height = .1)
plt.show()
```

```python
#Stripplot
sns.stripplot(x='day',y='total_bill',data=df,hue='sex',palette='Set1')
```

```python
#Catplot
sns.catplot(x='day',y='total_bill',data=df,kind='bar')
```

```python
#violinplot
sns.violinplot(x='day',y='total_bill', data = df, palette = 'Set1')
```

```python
#Swarmplot
sns.swarmplot(x='day',y='total_bill',data=df,hue='sex',palette='Set1')
```

```python
#Violinplot+swarmplot
sns.violinplot(x='tip',y='day',data=df, palette ='Set1')
sns.swarmplot(x='tip',y='day',data=df, color = 'black')
```

```python
#MATRIX PLOT
#HeatMap
flight = sns.load_dataset('flights')
```

```python
flight
```

```python
df
```

```python
tipscorr = df[['total_bill','tip','size']]
tipscorr
```

```python
cor = tipscorr.corr()
cor
```

```python
sns.heatmap(cor,annot=True)
```

```python
#ClusterMap
sns.clustermap(cor,cmap='coolwarm',standard_scale=1)
```

```python
#Heatmap for Pivot Table
pvtflight = flight.pivot_table(values='passengers',index='month',columns='year')
```

```python
pvtflight
```

```python
#Creating heatmap for pivot table
sns.heatmap(pvtflight)
```

```python
#Regression Plot
sns.lmplot(x='total_bill',y='tip',data=df, hue='sex',palette='Set1',markers=['o','v'],scatter_kws={'s':100})
```
