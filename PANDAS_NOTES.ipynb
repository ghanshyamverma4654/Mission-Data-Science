**MERGING**

```python
employees = pd.DataFrame({
    "employee_id":[1,2,3,4,5],
    "name":["John","Ann","Peter","Linda","Bob"],
    "department":["HR","IT","Finance","IT","HR"]
})
```

```python
employees
```

```python
salaries
```

```python
#MERGING
pd.merge(employees,salaries,on="employee_id")
```

```python
pd.merge(employees,salaries,on="employee_id",how="outer")
```

```python
pd.merge(employees,salaries,on="employee_id",how="inner")
```

```python
pd.merge(employees,salaries,on="employee_id",how="right")
```

```python
pd.merge(employees,salaries,on="employee_id",how="left")
```

**PIVOT TABLES AND CROSS TABLES**

```python
import numpy as np
import pandas as pd
```

```python
dp = {
    'Date': pd.date_range('2023-01-01', periods = 20),
    'Product': ['A','B','C','D'] * 5,
    'Region': ['East','West','North','South'] * 5,
    'Sales': np.random.randint(100,1000,20),
    'Units': np.random.randint(10,100,20),
    'Rep' : ['John','Mary','Bob','Alice'] * 5
}
```

```python
dpf = pd.DataFrame(dp)
dpf
```

```python
dpf['Month'] = dpf['Date'].dt.month_name()
```

```python
dpf
```

```python
dpf['Quarter'] = 'Q' + dpf['Date'].dt.quarter.astype(str)
```

```python
dpf
```

```python
#CREATING PIVOT TABLE
pd.pivot_table(dpf, values = 'Sales', index = 'Region',columns = 'Product')
```

```python
pd.pivot_table(dpf, values = 'Sales', index = 'Rep', columns = 'Product', aggfunc = 'sum')
```

```python
pd.pivot_table(dpf, values = ['Sales','Units'], index = 'Region', columns = 'Product', aggfunc = 'sum')
```

```python
dpf
```

**CROSS TABLES**

```python
pd.crosstab(dpf['Region'],dpf['Product'])
```

```python
pd.crosstab(dpf['Rep'],dpf['Region'])
```
