```python
#Advance array Manipulation
#Stacking: Matrices can't be stacked but Vectors can.
#Stacking Arrays(Vectors)
x = np.array([1,2,3])
y = np.array([4,5,6])
#Size must be same of both arrays
```

```python
np.vstack((x,y)) #Vertical Stack : Stacks arrays in up and down form
```

```python
np.hstack((x,y)) #Horizontal Stack : Stacks arrays in a single line
```

```python
#Column Stack : Stacks elements of each array into column
np.column_stack((x,y))
```

```python
#Splitting Arrays (Matrices)
z = np.arange(1,17).reshape(4,4)
z
```

```python
#Horizontal Split
np.hsplit(z,2)
```

```python
#Vertical Split
np.vsplit(z,4)
```
