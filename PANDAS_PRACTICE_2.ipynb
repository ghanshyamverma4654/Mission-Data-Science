```python
#Create a new column for episode count
def episode_count(txt):
  check = False
  data = ''
  for i in txt:
    if i == ')':
      check = False
      return data
    if check == True:
      data = data + i
    if i == '(':
      check = True
```

```python
anime['Episodes'] = anime['Title'].apply(episode_count)
```

```python
anime['Episodes'] = anime['Episodes'].str.replace(" eps","")
```

```python
anime['Episodes'] = anime['Episodes'].astype(int)
```

```python
#Function to find time stamp of an anime
def time_stamp(txt):
  check = False
  data = ''
  for i in range(len(txt)):
    if txt[i] == ')':
      for j in range(i+1,i+20):
        data += txt[j]

      return data
```

```python
#Adding a new column: Total Time
anime['Total Time'] = anime['Title'].apply(time_stamp)
```

```python
from dateutil.relativedelta import relativedelta
from datetime import datetime
```

```python
#Longest running anime
#Function
def duration(txt):
  try:
    start_str, end_str = txt.split(' - ')
    start_date = datetime.strptime(start_str, '%b %Y')
    end_date = datetime.strptime(end_str, '%b %Y')
    r = relativedelta(end_date, start_date)
    return r.years * 12 + r.months + 1
  except Exception as e:
    return e

```

```python
anime['Months'] = anime['Total Time'].apply(duration)
anime.head()
```
